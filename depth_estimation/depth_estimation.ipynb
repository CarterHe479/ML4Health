```python
import os

# Define paths
overhead_base = os.path.expanduser("~/Documents/nutrition5k_dataset/imagery/realsense_overhead/")
side_base = os.path.expanduser("~/Documents/nutrition5k_dataset/imagery/side_angles/")

# List folders
overhead_folders = set(f for f in os.listdir(overhead_base) if os.path.isdir(os.path.join(overhead_base, f)))
side_folders = set(f for f in os.listdir(side_base) if os.path.isdir(os.path.join(side_base, f)))

# Overlapping folders (present in both)
overlap_folders = overhead_folders & side_folders

# Count overhead RGB images
overhead_rgb_count = sum(
    1 for folder in overhead_folders
    if os.path.exists(os.path.join(overhead_base, folder, "rgb.png"))
)

# Count side images
side_img_count = 0
for folder in side_folders:
    folder_path = os.path.join(side_base, folder)
    side_imgs = [f for f in os.listdir(folder_path) if f.startswith("rgb_") and f.endswith(".png")]
    side_img_count += len(side_imgs)

# Count side images only for overlapping folders
side_img_count_overlap = 0
for folder in overlap_folders:
    folder_path = os.path.join(side_base, folder)
    side_imgs = [f for f in os.listdir(folder_path) if f.startswith("rgb_") and f.endswith(".png")]
    side_img_count_overlap += len(side_imgs)

# Print results
print(f"📦 Overhead folders: {len(overhead_folders)}")
print(f"📷 Overhead RGB images found: {overhead_rgb_count}")
print(f"📦 Side-angle folders: {len(side_folders)}")
print(f"🖼️  Total side RGB images: {side_img_count}")
print(f"🔗 Overlapping sample folders: {len(overlap_folders)}")
print(f"🖼️  Side RGB images in overlapping samples: {side_img_count_overlap}")
```
